```python
from pathlib import Path

import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
```

```python
INTERIM_DIR = Path("../data/interim").resolve()
RAW_DIR = Path("../data/raw").resolve()

A11Y_ROOT = (
    RAW_DIR
    / "02"
    / "Reduced-A11y-CUA"
)

instances = pd.read_csv(
    INTERIM_DIR / "a11y_instances.csv"
)

behavior = pd.read_csv(
    INTERIM_DIR / "a11y_behavior_features.csv"
)

trees = pd.read_csv(
    INTERIM_DIR / "a11y_tree_session_features.csv"
)

merged = pd.read_csv(
    INTERIM_DIR / "a11y_features_merged.csv"
)
```

```python
html_files = list(
    A11Y_ROOT.rglob("*.html")
)

print(
    "HTML encontrados:",
    len(html_files)
)
```

```python
def obtener_session_dir(
    archivo,
    raiz
):

    if archivo.parent.name == "web_logs":

        carpeta_sesion = (
            archivo.parent.parent
        )

    else:

        carpeta_sesion = (
            archivo.parent
        )

    return str(
        carpeta_sesion.relative_to(
            raiz
        )
    )
```

```python
html_sessions = {
    obtener_session_dir(
        archivo,
        A11Y_ROOT
    )
    for archivo in html_files
}

len(html_sessions)
```

```python
def tiene_nombre_control(
    elemento,
    soup
):

    # Texto visible
    texto = elemento.get_text(
        " ",
        strip=True
    )

    if texto:
        return True

    # aria-label
    aria = elemento.get(
        "aria-label"
    )

    if aria and aria.strip():
        return True

    # title
    title = elemento.get(
        "title"
    )

    if title and title.strip():
        return True

    # aria-labelledby
    labelledby = elemento.get(
        "aria-labelledby"
    )

    if labelledby:

        for identificador in (
            labelledby.split()
        ):

            etiqueta = soup.find(
                id=identificador
            )

            if (
                etiqueta
                and etiqueta.get_text(
                    " ",
                    strip=True
                )
            ):
                return True

    # <label for="...">
    element_id = elemento.get(
        "id"
    )

    if element_id:

        label = soup.find(
            "label",
            attrs={
                "for": element_id
            }
        )

        if (
            label
            and label.get_text(
                " ",
                strip=True
            )
        ):
            return True

    # <label><input ...></label>
    if elemento.find_parent(
        "label"
    ):
        return True

    return False
```

```python
def extraer_html_features(
    archivo,
    raiz
):

    with open(
        archivo,
        "r",
        encoding="utf-8",
        errors="ignore"
    ) as f:

        html = f.read()

    soup = BeautifulSoup(
        html,
        "lxml"
    )

    elementos = soup.find_all(True)

    # --------------------------------
    # CONTROLES INTERACTIVOS
    # --------------------------------

    controles = []

    controles.extend(
        soup.find_all("button")
    )

    controles.extend(
        soup.find_all(
            "a",
            href=True
        )
    )

    controles.extend(
        [
            x
            for x in soup.find_all(
                "input"
            )
            if x.get("type") != "hidden"
        ]
    )

    controles.extend(
        soup.find_all("select")
    )

    controles.extend(
        soup.find_all("textarea")
    )

    # Eliminar duplicados
    controles = list({
        id(x): x
        for x in controles
    }.values())

    controles_con_nombre = sum(
        tiene_nombre_control(
            control,
            soup
        )
        for control in controles
    )

    controles_sin_nombre = (
        len(controles)
        - controles_con_nombre
    )

    if len(controles) > 0:

        label_ratio = (
            controles_con_nombre
            / len(controles)
        )

    else:

        label_ratio = np.nan

    # --------------------------------
    # VALIDACIÓN
    # --------------------------------

    campos = (
        soup.find_all("input")
        + soup.find_all("select")
        + soup.find_all("textarea")
    )

    atributos_validacion = {
        "required",
        "pattern",
        "min",
        "max",
        "minlength",
        "maxlength"
    }

    campos_con_validacion = 0

    for campo in campos:

        if any(
            campo.has_attr(attr)
            for attr
            in atributos_validacion
        ):
            campos_con_validacion += 1

    # --------------------------------
    # PROGRESO
    # --------------------------------

    progress_present = int(
        soup.find("progress")
        is not None

        or

        soup.find(
            attrs={
                "role": "progressbar"
            }
        )
        is not None
    )

    # --------------------------------
    # REGIONES DE ESTADO
    # --------------------------------

    status_present = int(

        soup.find(
            attrs={
                "role": "status"
            }
        )
        is not None

        or

        soup.find(
            attrs={
                "aria-live": True
            }
        )
        is not None

        or

        soup.find(
            attrs={
                "role": "alert"
            }
        )
        is not None
    )

    # --------------------------------
    # BREADCRUMBS
    # --------------------------------

    breadcrumb_present = int(

        soup.find(
            attrs={
                "aria-label":
                    lambda x:
                    x
                    and
                    "breadcrumb"
                    in x.lower()
            }
        )
        is not None

        or

        soup.find(
            class_=
                lambda x:
                x
                and
                "breadcrumb"
                in str(x).lower()
        )
        is not None
    )

    # --------------------------------
    # DOM DEPTH
    # --------------------------------

    depths = []

    for element in elementos:

        depth = len(
            list(element.parents)
        )

        depths.append(depth)

    max_dom_depth = (
        max(depths)
        if depths
        else 0
    )

    return {

        "session_dir":
            obtener_session_dir(
                archivo,
                raiz
            ),

        "source_file":
            str(
                archivo.relative_to(
                    raiz
                )
            ),

        "html_element_count":
            len(elementos),

        "form_count":
            len(
                soup.find_all("form")
            ),

        "input_count":
            len(
                soup.find_all("input")
            ),

        "button_count":
            len(
                soup.find_all("button")
            ),

        "link_count":
            len(
                soup.find_all(
                    "a",
                    href=True
                )
            ),

        "nav_count":
            len(
                soup.find_all("nav")
            ),

        "interactive_controls_count":
            len(controles),

        "labeled_controls_count":
            controles_con_nombre,

        "unlabeled_controls_html_count":
            controles_sin_nombre,

        "control_label_ratio_html":
            label_ratio,

        "validated_fields_count":
            campos_con_validacion,

        "input_validation_present":
            int(
                campos_con_validacion > 0
            ),

        "progress_indicator_present":
            progress_present,

        "status_region_present":
            status_present,

        "breadcrumbs_present":
            breadcrumb_present,

        "max_dom_depth":
            max_dom_depth
    }
```

```python
html_records = []
html_errors = []

for i, archivo in enumerate(
    html_files,
    start=1
):

    try:

        features = (
            extraer_html_features(
                archivo,
                A11Y_ROOT
            )
        )

        html_records.append(
            features
        )

    except Exception as e:

        html_errors.append({

            "archivo":
                str(archivo),

            "error":
                str(e)
        })

    if i % 100 == 0:

        print(
            f"{i}/{len(html_files)}"
        )
```

```python
html_features = pd.DataFrame(
    html_records
)

print(
    html_features.shape
)

display(
    html_features.head()
)
```

```python
html_session_features = (
    html_features
    .groupby(
        "session_dir",
        as_index=False
    )
    .agg(

        html_snapshot_count=(
            "source_file",
            "count"
        ),

        mean_html_elements=(
            "html_element_count",
            "mean"
        ),

        max_html_elements=(
            "html_element_count",
            "max"
        ),

        mean_interactive_controls_html=(
            "interactive_controls_count",
            "mean"
        ),

        total_interactive_controls_html=(
            "interactive_controls_count",
            "sum"
        ),

        total_labeled_controls_html=(
            "labeled_controls_count",
            "sum"
        ),

        max_unlabeled_controls_html=(
            "unlabeled_controls_html_count",
            "max"
        ),

        input_validation_present=(
            "input_validation_present",
            "max"
        ),

        progress_indicator_present=(
            "progress_indicator_present",
            "max"
        ),

        status_region_present=(
            "status_region_present",
            "max"
        ),

        breadcrumbs_present=(
            "breadcrumbs_present",
            "max"
        ),

        max_dom_depth=(
            "max_dom_depth",
            "max"
        )
    )
)
```

```python
html_session_features[
    "control_label_ratio_html"
] = np.where(

    html_session_features[
        "total_interactive_controls_html"
    ] > 0,

    html_session_features[
        "total_labeled_controls_html"
    ]
    /
    html_session_features[
        "total_interactive_controls_html"
    ],

    np.nan
)
```

```python
a11y_features_enriched = (
    a11y_features
    .merge(
        html_session_features,
        on="session_dir",
        how="left",
        validate="one_to_one"
    )
)
```
