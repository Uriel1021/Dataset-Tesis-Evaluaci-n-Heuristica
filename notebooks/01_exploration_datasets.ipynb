```python
def cargar_metadata_a11y(archivos):

    registros = []

    for archivo in archivos:

        try:

            with open(
                archivo,
                "r",
                encoding="utf-8"
            ) as f:

                data = json.load(f)

            fila = pd.json_normalize(
                data,
                sep="."
            )

            fila["source_file"] = str(
                archivo.relative_to(a11y_root)
            )

            # Extraemos información de la ruta
            relativa = archivo.relative_to(a11y_root)

            if len(relativa.parts) > 0:
                fila["grupo"] = relativa.parts[0]

            registros.append(fila)

        except Exception as e:

            print(
                f"Error leyendo {archivo}: {e}"
            )

    return pd.concat(
        registros,
        ignore_index=True
    )
```

```python
def contar_unicos_seguro(serie):
    """
    Cuenta valores únicos incluso cuando la columna
    contiene listas o diccionarios.
    """

    serie = serie.dropna()

    if serie.empty:
        return 0

    def convertir_hashable(valor):

        if isinstance(valor, (list, dict)):
            return json.dumps(
                valor,
                sort_keys=True,
                ensure_ascii=False
            )

        if isinstance(valor, set):
            return tuple(
                sorted(valor)
            )

        return valor

    try:

        return (
            serie
            .map(convertir_hashable)
            .nunique()
        )

    except (TypeError, ValueError):

        return (
            serie
            .astype(str)
            .nunique()
        )


def resumen_dataframe(
    df,
    nombre="DataFrame"
):

    print("=" * 100)
    print(nombre)
    print("=" * 100)

    print(
        f"Filas: {df.shape[0]:,}"
    )

    print(
        f"Columnas: {df.shape[1]:,}"
    )

    resumen = pd.DataFrame({

        "variable":
            df.columns,

        "tipo":
            df.dtypes.astype(str).values,

        "nulos":
            df.isna().sum().values,

        "porcentaje_nulos":
            (
                df.isna().mean().values * 100
            ).round(2),

        "valores_unicos": [
            contar_unicos_seguro(
                df[col]
            )
            for col in df.columns
        ]

    })

    return resumen
```

```python
def cargar_interacciones_web_a11y(
    archivos,
    raiz
):
    registros = []
    errores = []

    for archivo in archivos:

        try:

            with open(
                archivo,
                "r",
                encoding="utf-8"
            ) as f:
                data = json.load(f)

            interactions = data.get(
                "interactions",
                []
            )

            if not isinstance(
                interactions,
                list
            ):
                continue

            if len(interactions) == 0:
                continue

            eventos = pd.json_normalize(
                interactions,
                sep="."
            )

            # Carpeta correspondiente a la tarea
            session_dir = str(
                archivo
                .parent
                .parent
                .relative_to(raiz)
            )

            eventos["session_dir"] = session_dir

            eventos["page_url"] = data.get(
                "url"
            )

            eventos["page_title"] = data.get(
                "title"
            )

            eventos["page_order"] = data.get(
                "order"
            )

            eventos["page_created_at"] = data.get(
                "created_at"
            )

            eventos["source_log"] = str(
                archivo.relative_to(raiz)
            )

            registros.append(
                eventos
            )

        except Exception as e:

            errores.append({
                "archivo": str(archivo),
                "error": str(e)
            })

    if not registros:
        return (
            pd.DataFrame(),
            pd.DataFrame(errores)
        )

    return (
        pd.concat(
            registros,
            ignore_index=True
        ),
        pd.DataFrame(errores)
    )
```

```python
def aplanar_arbol_a11y(
    nodo,
    profundidad=0,
    parent_id=None,
    filas=None
):

    if filas is None:
        filas = []

    node_id = len(filas)

    rect = nodo.get(
        "bounding_rectangle",
        {}
    ) or {}

    filas.append({
        "node_id": node_id,
        "parent_id": parent_id,
        "depth": profundidad,

        "name": nodo.get(
            "name"
        ),

        "control_type": nodo.get(
            "control_type"
        ),

        "automation_id": nodo.get(
            "automation_id"
        ),

        "left": rect.get(
            "left"
        ),

        "top": rect.get(
            "top"
        ),

        "right": rect.get(
            "right"
        ),

        "bottom": rect.get(
            "bottom"
        )
    })

    children = nodo.get(
        "children",
        []
    ) or []

    for child in children:

        if isinstance(child, dict):

            aplanar_arbol_a11y(
                child,
                profundidad + 1,
                node_id,
                filas
            )

    return filas
```

```python
INTERACTIVE_CONTROLS = {
    "ButtonControl",
    "EditControl",
    "ComboBoxControl",
    "CheckBoxControl",
    "RadioButtonControl",
    "HyperlinkControl",
    "MenuItemControl",
    "TabItemControl",
    "ListItemControl",
    "SliderControl",
    "SpinnerControl"
}
```

```python
def extraer_nodos_a11y(
    nodo,
    depth=0,
    filas=None
):

    if filas is None:
        filas = []

    filas.append({
        "name": nodo.get("name"),
        "control_type": nodo.get("control_type"),
        "automation_id": nodo.get("automation_id"),
        "depth": depth
    })

    children = nodo.get(
        "children",
        []
    ) or []

    for child in children:

        if isinstance(
            child,
            dict
        ):

            extraer_nodos_a11y(
                child,
                depth + 1,
                filas
            )

    return filas
```

```python
def resumir_arbol_a11y(
    archivo,
    raiz
):

    with open(
        archivo,
        "r",
        encoding="utf-8"
    ) as f:

        tree = json.load(f)

    nodos = extraer_nodos_a11y(
        tree
    )

    df_nodes = pd.DataFrame(
        nodos
    )

    interactive = df_nodes[
        df_nodes["control_type"]
        .isin(INTERACTIVE_CONTROLS)
    ].copy()

    nombres_validos = (
        interactive["name"]
        .fillna("")
        .astype(str)
        .str.strip()
        .ne("")
    )

    interactive_count = len(
        interactive
    )

    named_count = int(
        nombres_validos.sum()
    )

    unlabeled_count = (
        interactive_count
        - named_count
    )

    if interactive_count > 0:

        accessible_name_ratio = (
            named_count
            / interactive_count
        )

    else:

        accessible_name_ratio = np.nan

    relativa = (
        archivo
        .relative_to(raiz)
    )

    if archivo.parent.name == "web_logs":

        session_folder = (
            archivo.parent.parent
        )

    else:

        session_folder = (
            archivo.parent
        )

    session_dir = str(
        session_folder
        .relative_to(raiz)
    )

    return {
        "session_dir": session_dir,
        "source_file": str(relativa),

        "node_count": len(
            df_nodes
        ),

        "interactive_controls_count":
            interactive_count,

        "named_interactive_controls":
            named_count,

        "unlabeled_controls_count":
            unlabeled_count,

        "accessible_name_ratio":
            accessible_name_ratio,

        "max_a11y_tree_depth":
            df_nodes["depth"].max()
            if len(df_nodes)
            else np.nan
    }
```

```python
a11y_tree_records = []

errores_a11y = []

for i, archivo in enumerate(
    a11y_tree_files,
    start=1
):

    try:

        resultado = resumir_arbol_a11y(
            archivo,
            a11y_root
        )

        a11y_tree_records.append(
            resultado
        )

    except Exception as e:

        errores_a11y.append({
            "archivo": str(archivo),
            "error": str(e)
        })

    if i % 500 == 0:

        print(
            f"{i:,}/"
            f"{len(a11y_tree_files):,}"
        )
```
