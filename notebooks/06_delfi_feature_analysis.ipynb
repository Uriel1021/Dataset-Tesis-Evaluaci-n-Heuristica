```python
from pathlib import Path

import pandas as pd
import numpy as np

```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 150)
pd.set_option("display.max_colwidth", 150)

INTERIM_DIR = Path("../data/interim").resolve()

assign = pd.read_csv(
    INTERIM_DIR / "delfi_assign.csv"
)

quiz = pd.read_csv(
    INTERIM_DIR / "delfi_quiz.csv"
)

print("Assignments:", assign.shape)
print("Quiz:", quiz.shape)
```

```python
print("VARIABLES ASSIGNMENT")
print("=" * 80)

for i, col in enumerate(assign.columns, 1):
    print(f"{i:02d}. {col}")
```

```python
print("VARIABLES QUIZ")
print("=" * 80)

for i, col in enumerate(quiz.columns, 1):
    print(f"{i:02d}. {col}")
```

```python
assign_candidates = [
    "requiresubmissionstatement",
    "completionsubmit",
    "cutoffdate",
    "gradingduedate",
    "teamsubmission",
    "requireallteammemberssubmit",
    "attemptreopenmethod",
    "maxattempts",
    "preventsubmissionnotingroup"
]

assign_candidates = [
    col
    for col in assign_candidates
    if col in assign.columns
]

assign_candidates
```

```python
for col in assign_candidates:

    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)

    print(
        assign[col]
        .value_counts(
            dropna=False
        )
        .head(30)
    )
```

```python
quiz_candidates = [
    "timelimit",
    "overduehandling",
    "graceperiod",
    "preferredbehaviour",
    "canredoquestions",
    "attempts",

    "reviewattempt",
    "reviewcorrectness",
    "reviewmarks",
    "reviewspecificfeedback",
    "reviewgeneralfeedback",
    "reviewrightanswer",
    "reviewoverallfeedback",

    "questionsperpage",
    "navmethod",

    "completionattemptsexhausted",
    "completionpass"
]

quiz_candidates = [
    col
    for col in quiz_candidates
    if col in quiz.columns
]

quiz_candidates
```

```python
for col in quiz_candidates:

    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)

    print(
        quiz[col]
        .value_counts(
            dropna=False
        )
        .head(30)
    )
```

```python
delfi_mapping["existe_en_dataset"] = delfi_mapping.apply(
    lambda row:
        (
            row["variable_origen"] in assign.columns
            if row["fuente"] == "assignment"
            else
            row["variable_origen"] in quiz.columns
        ),
    axis=1
)

delfi_mapping[
    [
        "fuente",
        "variable_origen",
        "heuristica_relacionada",
        "variable_tesis_candidata",
        "estado",
        "existe_en_dataset"
    ]
]
```
