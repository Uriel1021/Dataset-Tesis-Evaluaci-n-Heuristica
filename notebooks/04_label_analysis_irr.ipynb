# Analizar las etiquetas heurísticas del repositorio 03

```python
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 150)
pd.set_option("display.max_colwidth", 150)

INTERIM_DIR = Path("../data/interim").resolve()
PROCESSED_DIR = Path("../data/processed").resolve()

irr = pd.read_csv(
    INTERIM_DIR / "irr_results.csv"
)

print("Dimensiones:", irr.shape)

display(irr.head())
```

```python
columnas_relevantes = [
    "Name",
    "Prototype Link",

    "H1 Compare",
    "H4 Compare",
    "H5 Compare",
    "H6 Compare",

    "H1. Overal Rating (4)",
    "H1",

    "H4 Overall Rating",
    "H4",

    "H5 Overall Rating",
    "H5",

    "H6 Overall Rating",
    "H6"
]

irr_selected = irr[
    columnas_relevantes
].copy()

display(
    irr_selected.head(20)
)
```

```python
resumen_irr = pd.DataFrame({
    "variable": irr_selected.columns,
    "tipo": irr_selected.dtypes.astype(str).values,
    "no_nulos": irr_selected.notna().sum().values,
    "nulos": irr_selected.isna().sum().values,
    "porcentaje_nulos": (
        irr_selected.isna().mean().values
        * 100
    ).round(2),

    "valores_unicos": [
        irr_selected[col].nunique(
            dropna=True
        )
        for col in irr_selected.columns
    ]
})

resumen_irr
```

```python
columnas_h = [
    "H1 Compare",
    "H4 Compare",
    "H5 Compare",
    "H6 Compare",

    "H1. Overal Rating (4)",
    "H1",

    "H4 Overall Rating",
    "H4",

    "H5 Overall Rating",
    "H5",

    "H6 Overall Rating",
    "H6"
]

for col in columnas_h:

    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)

    print(
        irr_selected[col]
        .value_counts(
            dropna=False
        )
        .sort_index()
    )
```

```python
compare_cols = [
    "H1 Compare",
    "H4 Compare",
    "H5 Compare",
    "H6 Compare"
]

rating_cols = [
    "H1. Overal Rating (4)",
    "H4 Overall Rating",
    "H5 Overall Rating",
    "H6 Overall Rating"
]

heuristic_cols = [
    "H1",
    "H4",
    "H5",
    "H6"
]
```

```python
irr_selected["has_compare"] = (
    irr_selected[
        compare_cols
    ]
    .notna()
    .any(axis=1)
)

irr_selected["has_rating"] = (
    irr_selected[
        rating_cols
    ]
    .notna()
    .any(axis=1)
)

irr_selected["has_h_value"] = (
    irr_selected[
        heuristic_cols
    ]
    .notna()
    .any(axis=1)
)
```

```python
irr_selected[
    [
        "has_compare",
        "has_rating",
        "has_h_value"
    ]
].value_counts()
```

```python
pd.crosstab(
    irr_selected["has_compare"],
    irr_selected["has_h_value"]
)
```

```python
irr_compare = irr_selected[
    irr_selected["has_compare"]
].copy()

print(
    "Filas Compare:",
    len(irr_compare)
)

display(
    irr_compare[
        [
            "Name",
            "H1 Compare",
            "H4 Compare",
            "H5 Compare",
            "H6 Compare"
        ]
    ]
)
```

```python
irr_ratings = irr_selected[
    irr_selected["has_rating"]
    |
    irr_selected["has_h_value"]
].copy()

print(
    "Filas Ratings:",
    len(irr_ratings)
)

display(
    irr_ratings[
        [
            "Name",

            "H1. Overal Rating (4)",
            "H1",

            "H4 Overall Rating",
            "H4",

            "H5 Overall Rating",
            "H5",

            "H6 Overall Rating",
            "H6"
        ]
    ].head(50)
)
```
