## 2. Protección contra sobrescritura accidental

El notebook solo reemplaza los registros cuando detecta que las plantillas existentes siguen esencialmente vacías. Si ya fueron editadas manualmente, se detiene.


````python
def registry_has_manual_content(
    path,
    text_columns
):

    if not path.exists():
        return False

    df = pd.read_csv(
        path
    )

    for col in text_columns:

        if col not in df.columns:
            continue

        values = (
            df[
                col
            ]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        if (
            values
            !=
            ""
        ).any():
            return True

    return False


site_manual_content = (
    registry_has_manual_content(
        SITE_PATH,
        [
            "site_name",
            "base_url",
            "domain",
            "educational_context"
        ]
    )
)

task_manual_content = (
    registry_has_manual_content(
        TASK_PATH,
        [
            "task_name",
            "task_description",
            "start_condition",
            "success_criterion"
        ]
    )
)

print(
    "Contenido manual en sites:",
    site_manual_content
)

print(
    "Contenido manual en tasks:",
    task_manual_content
)

if (
    site_manual_content
    or
    task_manual_content
):
    raise RuntimeError(
        "Los registros ya contienen información manual. "
        "No se sobrescribieron. Revisa los CSV antes de continuar."
    )
````

## 5. Validaciones antes de guardar

````python
assert len(
    site_registry_seed
) == 12

assert (
    site_registry_seed[
        "site_id"
    ]
    .duplicated()
    .sum()
    ==
    0
)

assert len(
    task_registry_seed
) == 24

assert (
    task_registry_seed[
        "task_id"
    ]
    .duplicated()
    .sum()
    ==
    0
)

assert set(
    task_registry_seed[
        "site_id"
    ]
).issubset(
    set(
        site_registry_seed[
            "site_id"
        ]
    )
)

tasks_per_site = (
    task_registry_seed
    .groupby(
        "site_id"
    )
    .size()
)

assert (
    tasks_per_site
    ==
    2
).all()

assert (
    site_registry_seed[
        "uses_personal_data"
    ]
    ==
    False
).all()

assert (
    site_registry_seed[
        "included"
    ]
    ==
    True
).all()

assert (
    task_registry_seed[
        "included"
    ]
    ==
    True
).all()

print(
    "12 sitios × 2 tareas = 24 tareas ✅"
)

print(
    "Sin IDs duplicados ✅"
)

print(
    "Sin uso planificado de datos personales reales ✅"
)
````
