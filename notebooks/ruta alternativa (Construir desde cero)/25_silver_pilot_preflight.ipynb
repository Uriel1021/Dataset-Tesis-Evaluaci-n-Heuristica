# 25 — Silver Pilot Preflight

## Objetivo

Convertir la especificación del notebook 24 en un plan operativo de piloto antes de recolectar datos reales.

El piloto se fija, como criterio operativo del proyecto, en:

- 12 sitios educativos,
- 2 tareas por sitio,
- 2 ejecuciones por tarea,
- 48 instancias totales.

Este notebook:

1. crea plantillas concretas para 12 sitios y 24 tareas;
2. valida que los registros estén completos antes de recolectar;
3. construye el `pilot_manifest.csv`;
4. valida exportaciones reales de la extensión cuando aparezcan;
5. comprueba unicidad, columnas X, rangos y faltantes;
6. prepara paquetes de anotación Silver;
7. genera un estado `READY / NOT_READY`.

No entrena modelos y no modifica `data/processed/`.

## 3. Diseño operativo del piloto

Para obtener 48 instancias sin tratar ejecuciones repetidas como sitios independientes:

\[
12\ 	ext{sitios}
	imes 2\ 	ext{tareas/sitio}
	imes 2\ 	ext{ejecuciones/tarea}
= 48\ 	ext{instancias}
\]

Las repeticiones sirven para comprobar estabilidad de captura, no para inflar artificialmente el número de sitios.


````python
PILOT_SITES = 12
TASKS_PER_SITE = 2
EXECUTIONS_PER_TASK = 2

EXPECTED_TASKS = (
    PILOT_SITES
    *
    TASKS_PER_SITE
)

EXPECTED_INSTANCES = (
    EXPECTED_TASKS
    *
    EXECUTIONS_PER_TASK
)

print("Sitios:", PILOT_SITES)
print("Tareas:", EXPECTED_TASKS)
print("Instancias:", EXPECTED_INSTANCES)

assert EXPECTED_INSTANCES == 48
````

### Qué llenar en `site_registry_pilot.csv`

Para cada uno de los 12 sitios:

- `included`: `True` solo cuando el sitio haya sido revisado y aceptado.

No usar cuentas ajenas, datos personales reales ni flujos de pago.

- mantener `NaN` en `applicable_H*` cuando aún no se haya determinado la aplicabilidad.

## 6. Validar registro de sitios

````python
def normalize_bool_series(
    series
):

    mapping = {
        True: True,
        False: False,
        "true": True,
        "false": False,
        "True": True,
        "False": False,
        "1": True,
        "0": False,
        1: True,
        0: False
    }

    return series.map(
        mapping
    )


site_registry = pd.read_csv(
    site_template_path
)

for col in [
    "included",
    "requires_account",
    "uses_personal_data"
]:
    site_registry[
        col
    ] = normalize_bool_series(
        site_registry[
            col
        ]
    )

included_sites = site_registry[
    site_registry[
        "included"
    ]
    ==
    True
].copy()

site_checks = []

site_checks.append({
    "check":
        "TOTAL_INCLUDED_SITES",

    "passed":
        len(
            included_sites
        )
        ==
        PILOT_SITES,

    "observed":
        len(
            included_sites
        ),

    "expected":
        PILOT_SITES
})

site_checks.append({
    "check":
        "UNIQUE_SITE_ID",

    "passed":
        included_sites[
            "site_id"
        ].duplicated().sum()
        ==
        0,

    "observed":
        int(
            included_sites[
                "site_id"
            ].duplicated().sum()
        ),

    "expected":
        0
})

required_site_text = [
    "site_name",
    "base_url",
    "domain",
    "educational_context",
    "access_type"
]

for col in required_site_text:

    missing = (
        included_sites[
            col
        ]
        .fillna("")
        .astype(str)
        .str.strip()
        ==
        ""
    ).sum()

    site_checks.append({
        "check":
            f"SITE_REQUIRED_{col}",

        "passed":
            int(
                missing
            )
            ==
            0,

        "observed":
            int(
                missing
            ),

        "expected":
            0
    })

personal_data_true = (
    included_sites[
        "uses_personal_data"
    ]
    ==
    True
).sum()

site_checks.append({
    "check":
        "NO_PERSONAL_DATA",

    "passed":
        int(
            personal_data_true
        )
        ==
        0,

    "observed":
        int(
            personal_data_true
        ),

    "expected":
        0
})

site_validation = pd.DataFrame(
    site_checks
)

display(
    site_validation
)
````

## 7. Validar registro de tareas

````python
task_registry = pd.read_csv(
    task_template_path
)

task_registry[
    "included"
] = normalize_bool_series(
    task_registry[
        "included"
    ]
)

included_tasks = task_registry[
    task_registry[
        "included"
    ]
    ==
    True
].copy()

task_checks = []

task_checks.append({
    "check":
        "TOTAL_INCLUDED_TASKS",

    "passed":
        len(
            included_tasks
        )
        ==
        EXPECTED_TASKS,

    "observed":
        len(
            included_tasks
        ),

    "expected":
        EXPECTED_TASKS
})

task_checks.append({
    "check":
        "UNIQUE_TASK_ID",

    "passed":
        included_tasks[
            "task_id"
        ].duplicated().sum()
        ==
        0,

    "observed":
        int(
            included_tasks[
                "task_id"
            ].duplicated().sum()
        ),

    "expected":
        0
})

unknown_sites = (
    set(
        included_tasks[
            "site_id"
        ]
    )
    -
    set(
        included_sites[
            "site_id"
        ]
    )
)

task_checks.append({
    "check":
        "TASK_SITE_FOREIGN_KEY",

    "passed":
        len(
            unknown_sites
        )
        ==
        0,

    "observed":
        str(
            sorted(
                unknown_sites
            )
        ),

    "expected":
        "[]"
})

required_task_text = [
    "task_name",
    "task_description",
    "start_condition",
    "success_criterion"
]

for col in required_task_text:

    missing = (
        included_tasks[
            col
        ]
        .fillna("")
        .astype(str)
        .str.strip()
        ==
        ""
    ).sum()

    task_checks.append({
        "check":
            f"TASK_REQUIRED_{col}",

        "passed":
            int(
                missing
            )
            ==
            0,

        "observed":
            int(
                missing
            ),

        "expected":
            0
    })

tasks_per_site = (
    included_tasks
    .groupby(
        "site_id"
    )
    .size()
)

bad_task_counts = (
    tasks_per_site[
        tasks_per_site
        !=
        TASKS_PER_SITE
    ]
)

task_checks.append({
    "check":
        "TASKS_PER_SITE",

    "passed":
        len(
            bad_task_counts
        )
        ==
        0
        and
        len(
            tasks_per_site
        )
        ==
        PILOT_SITES,

    "observed":
        str(
            tasks_per_site
            .to_dict()
        ),

    "expected":
        (
            f"{TASKS_PER_SITE} por sitio"
        )
})

task_validation = pd.DataFrame(
    task_checks
)

display(
    task_validation
)
````

## 8. Construir manifiesto de 48 instancias

````python
registries_ready = (
    site_validation[
        "passed"
    ].all()
    and
    task_validation[
        "passed"
    ].all()
)

pilot_manifest_path = (
    META_DIR
    / "pilot_manifest.csv"
)

if registries_ready:

    manifest_rows = []

    for _, task in (
        included_tasks.iterrows()
    ):

        for execution_index in range(
            1,
            EXECUTIONS_PER_TASK + 1
        ):

            execution_id = (
                f"exec_{execution_index:02d}"
            )

            instance_id = (
                f"{task['site_id']}"
                f"__{task['task_id'].split('__')[-1]}"
                f"__{execution_id}"
            )

            manifest_rows.append({
                "instance_id":
                    instance_id,

                "site_id":
                    task[
                        "site_id"
                    ],

                "task_id":
                    task[
                        "task_id"
                    ],

                "execution_id":
                    execution_id,

                "collection_status":
                    "PLANNED",

                "extension_export_path":
                    (
                        "data/silver/raw/"
                        "extension_exports/"
                        f"{instance_id}.csv"
                    ),

                "evidence_dir":
                    (
                        "data/silver/raw/"
                        "evidence/"
                        f"{instance_id}"
                    ),

                "label_runs_required":
                    3
            })

    pilot_manifest = pd.DataFrame(
        manifest_rows
    )

    assert len(
        pilot_manifest
    ) == EXPECTED_INSTANCES

    assert (
        pilot_manifest[
            "instance_id"
        ]
        .duplicated()
        .sum()
        ==
        0
    )

    pilot_manifest.to_csv(
        pilot_manifest_path,
        index=False
    )

else:

    pilot_manifest = pd.DataFrame(
        columns=[
            "instance_id",
            "site_id",
            "task_id",
            "execution_id",
            "collection_status",
            "extension_export_path",
            "evidence_dir",
            "label_runs_required"
        ]
    )

    pilot_manifest.to_csv(
        pilot_manifest_path,
        index=False
    )

print(
    "Registries ready:",
    registries_ready
)

print(
    "Pilot manifest:",
    pilot_manifest.shape
)

display(
    pilot_manifest.head(
        20
    )
)
````

## 11. Función de validación de export

````python
def validate_extension_export(
    path
):

    result = {
        "file":
            str(
                path
            ),

        "read_ok":
            False,

        "rows":
            np.nan,

        "missing_columns":
            None,

        "extra_columns":
            None,

        "duplicate_instance_id":
            np.nan,

        "ratio_violations":
            0,

        "binary_violations":
            0,

        "negative_violations":
            0,

        "valid":
            False,

        "error":
            None
    }

    try:

        df = pd.read_csv(
            path
        )

    except Exception as e:

        result[
            "error"
        ] = str(e)

        return result

    result[
        "read_ok"
    ] = True

    result[
        "rows"
    ] = len(
        df
    )

    missing = sorted(
        set(
            REQUIRED_EXPORT_COLUMNS
        )
        -
        set(
            df.columns
        )
    )

    extra = sorted(
        set(
            df.columns
        )
        -
        set(
            REQUIRED_EXPORT_COLUMNS
        )
    )

    result[
        "missing_columns"
    ] = " | ".join(
        missing
    )

    result[
        "extra_columns"
    ] = " | ".join(
        extra
    )

    if (
        "instance_id"
        in df.columns
    ):

        result[
            "duplicate_instance_id"
        ] = int(
            df[
                "instance_id"
            ]
            .duplicated()
            .sum()
        )

    ratio_violations = 0

    for col in RATIO_COLUMNS:

        if col not in df.columns:
            continue

        values = pd.to_numeric(
            df[
                col
            ],
            errors="coerce"
        )

        ratio_violations += int(
            (
                (
                    values
                    <
                    0
                )
                |
                (
                    values
                    >
                    1
                )
            ).sum()
        )

    binary_violations = 0

    for col in BINARY_COLUMNS:

        if col not in df.columns:
            continue

        values = pd.to_numeric(
            df[
                col
            ],
            errors="coerce"
        )

        binary_violations += int(
            (
                values
                .notna()
                &
                ~values.isin(
                    [
                        0,
                        1
                    ]
                )
            ).sum()
        )

    negative_violations = 0

    for col in (
        COUNT_COLUMNS
        +
        NONNEGATIVE_COLUMNS
    ):

        if col not in df.columns:
            continue

        values = pd.to_numeric(
            df[
                col
            ],
            errors="coerce"
        )

        negative_violations += int(
            (
                values
                <
                0
            ).sum()
        )

    result[
        "ratio_violations"
    ] = ratio_violations

    result[
        "binary_violations"
    ] = binary_violations

    result[
        "negative_violations"
    ] = negative_violations

    result[
        "valid"
    ] = (
        result[
            "read_ok"
        ]
        and
        len(
            missing
        )
        ==
        0
        and
        (
            result[
                "duplicate_instance_id"
            ]
            in [
                0,
                np.nan
            ]
            or
            pd.isna(
                result[
                    "duplicate_instance_id"
                ]
            )
        )
        and
        ratio_violations
        ==
        0
        and
        binary_violations
        ==
        0
        and
        negative_violations
        ==
        0
    )

    return result
````

## 15. Crear paquetes de anotación

````python
annotation_package_records = []

if not pilot_manifest.empty:

    task_lookup = (
        included_tasks
        .set_index(
            "task_id"
        )
        .to_dict(
            orient="index"
        )
    )

    for _, row in (
        pilot_manifest.iterrows()
    ):

        instance_id = row[
            "instance_id"
        ]

        task_info = task_lookup.get(
            row[
                "task_id"
            ],
            {}
        )

        package = {
            "instance_id":
                instance_id,

            "site_id":
                row[
                    "site_id"
                ],

            "task_id":
                row[
                    "task_id"
                ],

            "execution_id":
                row[
                    "execution_id"
                ],

            "task_description":
                task_info.get(
                    "task_description"
                ),

            "start_condition":
                task_info.get(
                    "start_condition"
                ),

            "success_criterion":
                task_info.get(
                    "success_criterion"
                ),

            "applicable_H1":
                task_info.get(
                    "applicable_H1"
                ),

            "applicable_H4":
                task_info.get(
                    "applicable_H4"
                ),

            "applicable_H5":
                task_info.get(
                    "applicable_H5"
                ),

            "applicable_H6":
                task_info.get(
                    "applicable_H6"
                ),

            "evidence_dir":
                row[
                    "evidence_dir"
                ],

            "x_values_included":
                False,

            "model_prediction_included":
                False,

            "annotation_runs_required":
                3,

            "prompt_version":
                "silver_v1"
        }

        package_path = (
            ANNOTATION_PACKAGE_DIR
            /
            f"{instance_id}.json"
        )

        package_path.write_text(
            json.dumps(
                package,
                ensure_ascii=False,
                indent=2,
                default=str
            ),
            encoding="utf-8"
        )

        annotation_package_records.append({
            "instance_id":
                instance_id,

            "package_path":
                str(
                    package_path.relative_to(
                        PROJECT_DIR
                    )
                ),

            "created":
                package_path.exists()
        })

annotation_packages = pd.DataFrame(
    annotation_package_records
)

display(
    annotation_packages.head(
        20
    )
)
````
