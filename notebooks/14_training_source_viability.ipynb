```python
from pathlib import Path

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
pd.set_option("display.max_colwidth", 300)

PROJECT_DIR = Path("..").resolve()

REPORTS_DIR = (
    PROJECT_DIR
    / "reports"
    / "tables"
)

INTERIM_DIR = (
    PROJECT_DIR
    / "data"
    / "public_v2"
    / "interim"
)

REPORTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
label_registry = pd.read_csv(
    REPORTS_DIR
    / "dataset_label_registry.csv"
)

source_feasibility = pd.read_csv(
    REPORTS_DIR
    / "source_feature_feasibility.csv"
)

catching_final = pd.read_csv(
    REPORTS_DIR
    / "catching_repository_final.csv"
)

catching_targets = pd.read_csv(
    INTERIM_DIR
    / "catching_targets.csv"
)

print(
    "Label registry:",
    label_registry.shape
)

print(
    "Source feasibility:",
    source_feasibility.shape
)

print(
    "Catching repositories:",
    catching_final.shape
)

print(
    "Catching targets:",
    catching_targets.shape
)
```

```python
catching_repository_counts = (
    catching_final[
        "repository_status"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "repository_status"
    )
    .reset_index(
        name="count"
    )
)

catching_repository_counts
```

```python
n_strong = (
    catching_final[
        "repository_status"
    ]
    ==
    "CONFIRMED_STRONG"
).sum()

n_moderate = (
    catching_final[
        "repository_status"
    ]
    ==
    "PROVISIONAL_MODERATE"
).sum()

n_ambiguous = (
    catching_final[
        "repository_status"
    ]
    .astype(str)
    .str.startswith(
        "AMBIGUOUS"
    )
).sum()

n_unresolved = (
    catching_final[
        "repository_status"
    ]
    ==
    "UNRESOLVED"
).sum()

print(
    "Strong:",
    n_strong
)

print(
    "Moderate:",
    n_moderate
)

print(
    "Ambiguous:",
    n_ambiguous
)

print(
    "Unresolved:",
    n_unresolved
)
```
