```python
target_cols = [
    "H1_violation",
    "H4_violation",
    "H5_violation",
    "H6_violation"
]

balance_records = []

for target in target_cols:

    y = pd.to_numeric(
        catching_targets[target],
        errors="coerce"
    )

    n_positive = int(
        (y == 1).sum()
    )

    n_negative = int(
        (y == 0).sum()
    )

    n_missing = int(
        y.isna().sum()
    )

    n_valid = (
        n_positive
        +
        n_negative
    )

    prevalence = (
        n_positive / n_valid
        if n_valid > 0
        else np.nan
    )

    balance_records.append({
        "target": target,
        "positive": n_positive,
        "negative": n_negative,
        "missing": n_missing,
        "valid": n_valid,
        "positive_ratio": prevalence
    })

catching_balance = pd.DataFrame(
    balance_records
)

catching_balance
```
