```python
def normalize_repo_name(value):

    if pd.isna(value):
        return ""

    value = str(value)

    value = unicodedata.normalize(
        "NFKD",
        value
    )

    value = (
        value
        .encode(
            "ascii",
            "ignore"
        )
        .decode("ascii")
    )

    value = value.lower()

    value = re.sub(
        r"[^a-z0-9]+",
        "-",
        value
    )

    value = value.strip("-")

    return value
```

```python
sites[
    "normalized_site"
] = sites[
    "site"
].apply(
    normalize_repo_name
)

display(
    sites[
        [
            "site",
            "normalized_site"
        ]
    ]
)
```

```python
GITHUB_TOKEN = os.getenv(
    "GITHUB_TOKEN"
)

headers = {
    "Accept":
        "application/vnd.github+json",

    "X-GitHub-Api-Version":
        "2022-11-28"
}

if GITHUB_TOKEN:

    headers[
        "Authorization"
    ] = (
        f"Bearer {GITHUB_TOKEN}"
    )

print(
    "GitHub token:",
    "AVAILABLE"
    if GITHUB_TOKEN
    else "NOT SET"
)
```

```python
def search_github_repositories(
    query,
    per_page=10,
    max_retries=3
):

    url = (
        "https://api.github.com/"
        "search/repositories"
    )

    for attempt in range(
        max_retries
    ):

        response = requests.get(
            url,
            headers=headers,
            params={
                "q": query,
                "per_page": per_page
            },
            timeout=30
        )

        # Búsqueda correcta
        if response.status_code == 200:

            payload = response.json()

            return {
                "ok": True,
                "status_code": 200,
                "error": None,
                "items": payload.get(
                    "items",
                    []
                )
            }

        # Rate limit
        if response.status_code in {
            403,
            429
        }:

            remaining = response.headers.get(
                "X-RateLimit-Remaining"
            )

            reset = response.headers.get(
                "X-RateLimit-Reset"
            )

            print(
                f"Rate limit para '{query}'. "
                f"remaining={remaining}, "
                f"reset={reset}"
            )

            if reset is not None:

                wait_seconds = max(
                    int(reset)
                    - int(time.time())
                    + 2,
                    2
                )

            else:

                wait_seconds = 65

            print(
                "Esperando",
                wait_seconds,
                "segundos..."
            )

            time.sleep(
                wait_seconds
            )

            continue

        # Otro error
        return {
            "ok": False,
            "status_code":
                response.status_code,
            "error":
                response.text[:500],
            "items": []
        }

    return {
        "ok": False,
        "status_code":
            response.status_code,
        "error":
            "Maximum retries reached",
        "items": []
    }
```

```python
candidate_records = []
search_errors = []

for i, row in sites.iterrows():

    site = row["site"]

    print(
        f"[{i + 1:02d}/{len(sites)}]",
        site
    )

    result = (
        search_github_repositories(
            site,
            per_page=10
        )
    )

    if not result["ok"]:

        search_errors.append({
            "site":
                site,

            "status_code":
                result[
                    "status_code"
                ],

            "error":
                result[
                    "error"
                ]
        })

        continue

    for rank, repo in enumerate(
        result["items"],
        start=1
    ):

        candidate_records.append({

            "site":
                site,

            "normalized_site":
                row[
                    "normalized_site"
                ],

            "rank":
                rank,

            "repo_full_name":
                repo.get(
                    "full_name"
                ),

            "repo_name":
                repo.get(
                    "name"
                ),

            "repo_normalized":
                normalize_repo_name(
                    repo.get(
                        "name"
                    )
                ),

            "html_url":
                repo.get(
                    "html_url"
                ),

            "clone_url":
                repo.get(
                    "clone_url"
                ),

            "default_branch":
                repo.get(
                    "default_branch"
                ),

            "description":
                repo.get(
                    "description"
                ),

            "language":
                repo.get(
                    "language"
                ),

            "stars":
                repo.get(
                    "stargazers_count"
                ),

            "fork":
                repo.get(
                    "fork"
                ),

            "archived":
                repo.get(
                    "archived"
                ),

            "license_key":
                (
                    repo.get(
                        "license"
                    )
                    or {}
                ).get(
                    "key"
                ),

            "license_name":
                (
                    repo.get(
                        "license"
                    )
                    or {}
                ).get(
                    "name"
                )
        })

    # Separación preventiva entre búsquedas
    if not GITHUB_TOKEN:
        time.sleep(7)
```

```python
candidates = pd.DataFrame(
    candidate_records
)

candidates[
    "exact_name_match"
] = (
    candidates[
        "normalized_site"
    ]
    ==
    candidates[
        "repo_normalized"
    ]
)

print(
    "Candidates:",
    candidates.shape
)

print(
    "Sitios buscados correctamente:",
    candidates[
        "site"
    ].nunique()
)

print(
    "Errores:",
    len(search_errors)
)
```

```python
searched_ok = (
    len(sites)
    -
    len(search_errors)
)

print(
    "Búsquedas completadas:",
    searched_ok,
    "/",
    len(sites)
)

print(
    "Búsquedas fallidas:",
    len(search_errors)
)
```

```python
recovery_status = (
    sites[
        [
            "site",
            "normalized_site"
        ]
    ]
    .copy()
)

recovery_status[
    "candidate_count"
] = (
    recovery_status[
        "site"
    ]
    .map(
        candidates
        .groupby("site")
        .size()
    )
    .fillna(0)
    .astype(int)
)

recovery_status[
    "exact_match_count"
] = (
    recovery_status[
        "site"
    ]
    .map(
        exact_candidates
        .groupby("site")
        .size()
    )
    .fillna(0)
    .astype(int)
)

recovery_status[
    "recovery_status"
] = np.where(
    recovery_status[
        "exact_match_count"
    ] > 0,
    "CANDIDATE",
    "UNRESOLVED"
)

recovery_status[
    "confirmed_repo"
] = None

recovery_status[
    "evidence"
] = None
```

```python
exact = candidates[
    candidates[
        "exact_name_match"
    ] == True
].copy()

display(
    exact[
        [
            "site",
            "repo_full_name",
            "html_url",
            "language",
            "license_name",
            "archived",
            "fork",
            "stars"
        ]
    ]
    .sort_values(
        [
            "site",
            "stars"
        ],
        ascending=[
            True,
            False
        ]
    )
)
```

```python
exact_summary = (
    exact
    .groupby("site")
    .agg(
        exact_repositories=(
            "repo_full_name",
            "nunique"
        ),

        repositories=(
            "repo_full_name",
            lambda x:
                " | ".join(
                    sorted(
                        set(
                            x.dropna()
                        )
                    )
                )
        )
    )
    .reset_index()
)

display(
    exact_summary
)
```

```python
recovery_review = (
    recovery
    .merge(
        exact_summary,
        on="site",
        how="left"
    )
)

recovery_review[
    "exact_repositories"
] = (
    recovery_review[
        "exact_repositories"
    ]
    .fillna(0)
    .astype(int)
)
```

```python
def preliminary_recovery_decision(row):

    n = row[
        "exact_repositories"
    ]

    if n == 0:
        return "UNRESOLVED"

    if n == 1:
        return "REVIEW_SINGLE_CANDIDATE"

    return "REVIEW_MULTIPLE_CANDIDATES"


recovery_review[
    "review_status"
] = recovery_review.apply(
    preliminary_recovery_decision,
    axis=1
)
```
