```python
CATCHING_DIR = (
    PUBLIC_RAW
    / "04_catching_ux_flaws"
    / "results"
)

records = []

for path in sorted(
    CATCHING_DIR.rglob("*.json")
):

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        data = json.load(f)

    site = path.parent.name

    evaluation = (
        path.stem
        .lower()
        .replace(" ", "")
    )

    for item in data:

        records.append({
            "site": site,
            "evaluation": evaluation,
            "heuristic_raw": item.get("Heuristic"),
            "issue_found": item.get("IssueFound"),
            "severity": item.get("SeverityRating"),
            "source_file": str(
                path.relative_to(PUBLIC_RAW)
            )
        })

catching_raw = pd.DataFrame(records)

print(catching_raw.shape)
```

```python
def normalize_nielsen(value):

    if pd.isna(value):
        return None

    text = str(value).strip().lower()

    text = re.sub(
        r"^\d+\.\s*",
        "",
        text
    )

    mapping = {
        "visibility of system status": "H1",
        "consistency and standards": "H4",
        "error prevention": "H5",
        "recognition rather than recall": "H6"
    }

    if text in mapping:
        return mapping[text]

    m = re.fullmatch(
        r"heuristic\s*(\d+)",
        text
    )

    if m:

        number = int(m.group(1))

        if number in {1, 4, 5, 6}:
            return f"H{number}"

    return None
```

```python
catching_raw[
    "heuristic"
] = catching_raw[
    "heuristic_raw"
].apply(
    normalize_nielsen
)

catching_selected = catching_raw[
    catching_raw["heuristic"].notna()
].copy()

print(catching_selected.shape)

display(
    pd.crosstab(
        catching_selected["heuristic"],
        catching_selected["issue_found"]
    )
)
```

```python
catching_selected[
    "violation"
] = (
    catching_selected[
        "issue_found"
    ]
    .astype("boolean")
    .astype("Int8")
)
```

```python
catching_labels_evaluator = (
    catching_selected[
        [
            "site",
            "evaluation",
            "heuristic",
            "violation",
            "severity"
        ]
    ]
    .copy()
)

print(
    catching_labels_evaluator.shape
)
```

```python
def majority_consensus(series):

    values = (
        pd.to_numeric(
            series,
            errors="coerce"
        )
        .dropna()
        .astype(int)
    )

    if len(values) == 0:
        return pd.NA

    positives = int(
        (values == 1).sum()
    )

    negatives = int(
        (values == 0).sum()
    )

    if positives > negatives:
        return 1

    if negatives > positives:
        return 0

    # Empate: no inventar etiqueta
    return pd.NA
```

```python
catching_consensus_long = (
    catching_labels_evaluator
    .groupby(
        [
            "site",
            "heuristic"
        ]
    )
    .agg(
        n_votes=(
            "violation",
            "count"
        ),

        positive_votes=(
            "violation",
            "sum"
        ),

        consensus=(
            "violation",
            majority_consensus
        ),

        max_severity=(
            "severity",
            "max"
        )
    )
    .reset_index()
)

display(
    catching_consensus_long.head(30)
)
```
