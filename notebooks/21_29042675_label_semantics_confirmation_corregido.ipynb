## 3. Corregir la modalidad de las fuentes

En el notebook 20 una URL que no era Figma/Behance quedó provisionalmente como `OTHER_SOURCE`. Aquí no se interpreta automáticamente como fuente web. Una URL SharePoint de imagen/documento sigue siendo un artefacto, no HTML/CSS/JS de la interfaz evaluada.


```python
def source_domain(value):

    if pd.isna(value):
        return None

    try:
        return (
            urlparse(str(value))
            .netloc
            .lower()
            .replace("www.", "")
        )
    except Exception:
        return None


def classify_source_modality(value):

    domain = source_domain(value)

    if domain is None:
        return "MISSING"

    if "figma.com" in domain:
        return "FIGMA_PROTOTYPE"

    if "behance.net" in domain:
        return "BEHANCE_PRESENTATION"

    if "sharepoint.com" in domain:
        return "SHAREPOINT_ARTIFACT"

    return "OTHER_ARTIFACT"


metadata_source_audit = metadata.copy()

metadata_source_audit[
    "source_domain"
] = metadata_source_audit[
    "Scource URL"
].apply(source_domain)

metadata_source_audit[
    "source_modality"
] = metadata_source_audit[
    "Scource URL"
].apply(classify_source_modality)

metadata_source_audit[
    "canonical_web_X_ready"
] = False

print(
    metadata_source_audit[
        "source_modality"
    ].value_counts(dropna=False)
)

display(metadata_source_audit)
```

## 4. Extraer texto de DOCX

```python
def extract_docx_text(path):

    try:
        with zipfile.ZipFile(path) as z:
            xml = z.read("word/document.xml")

        root = ET.fromstring(xml)

        chunks = []

        for element in root.iter():

            if (
                element.tag.endswith("}t")
                and
                element.text
            ):
                chunks.append(element.text)

        return " ".join(chunks)

    except Exception as e:

        print("DOCX error:", path.name, e)
        return ""
```

## 5. Extraer texto básico de RTF

```python
def rtf_to_text(raw_text):

    # Caracteres hexadecimales RTF: \'hh
    def hex_replace(match):
        try:
            return bytes.fromhex(
                match.group(1)
            ).decode(
                "cp1252",
                errors="ignore"
            )
        except Exception:
            return ""

    text = re.sub(
        r"\\'([0-9a-fA-F]{2})",
        hex_replace,
        raw_text
    )

    # Unicode RTF: \uN?
    def unicode_replace(match):
        try:
            value = int(match.group(1))
            if value < 0:
                value += 65536
            return chr(value)
        except Exception:
            return ""

    text = re.sub(
        r"\\u(-?\d+)\??",
        unicode_replace,
        text
    )

    # Separadores comunes
    text = re.sub(
        r"\\par(?:d)?\b",
        "\n",
        text
    )

    text = re.sub(
        r"\\tab\b",
        "\t",
        text
    )

    # Eliminar destinos RTF y control words
    text = re.sub(
        r"\{\\\*[^{}]*\}",
        " ",
        text
    )

    text = re.sub(
        r"\\[a-zA-Z]+-?\d* ?",
        " ",
        text
    )

    text = text.replace("{", " ").replace("}", " ")

    text = re.sub(
        r"[ \t]+",
        " ",
        text
    )

    text = re.sub(
        r"\n\s*\n+",
        "\n",
        text
    )

    return html.unescape(
        text
    ).strip()


rtf_texts = {}

for filename in [
    "README.rtf",
    "EvaluationTemplate.rtf"
]:

    path = VT_DIR / filename

    if not path.exists():
        print(filename, "❌")
        continue

    raw = path.read_text(
        encoding="latin-1",
        errors="ignore"
    )

    rtf_texts[
        filename
    ] = rtf_to_text(raw)

    print(
        "\n",
        "=" * 100
    )

    print(filename)

    print(
        rtf_texts[
            filename
        ][:6000]
    )
```

## 6. Buscar evidencia semántica en los RTF

```python
RTF_TERMS = [
    "compare",
    "comparison",
    "inter-rater",
    "agreement",
    "overall rating",
    "severity",
    "0 to 4",
    "0-4",
    "evaluator",
    "evaluation 1",
    "evaluation 2",
    "chatgpt",
    "gemini",
    "heuristic"
]

rtf_evidence_records = []

for filename, text in rtf_texts.items():

    lower = text.lower()

    for term in RTF_TERMS:

        start_search = 0

        while True:

            idx = lower.find(
                term.lower(),
                start_search
            )

            if idx == -1:
                break

            start = max(
                0,
                idx - 300
            )

            end = min(
                len(text),
                idx + 700
            )

            rtf_evidence_records.append({
                "file":
                    filename,

                "term":
                    term,

                "context":
                    text[
                        start:end
                    ]
            })

            start_search = idx + len(term)

rtf_evidence = pd.DataFrame(
    rtf_evidence_records,
    columns=[
        "file",
        "term",
        "context"
    ]
)

print(
    "RTF evidence rows:",
    len(rtf_evidence)
)

display(
    rtf_evidence
)
```

## 8. Inventario de evaluaciones LLM

La unidad de evidencia será un archivo DOCX de evaluación. No se asumirá que el nombre del archivo equivale al nombre del prototipo.


```python
docx_files = sorted(
    [
        path
        for path
        in (
            VT_DIR
            / "LLM_Evaluations"
        ).rglob("*.docx")
        if path.is_file()
    ]
)

print(
    "DOCX encontrados:",
    len(docx_files)
)
```

```python
def infer_evaluation_round(path):

    text = str(path)

    match = re.search(
        r"LLM Evaluation\s*(\d+)",
        text,
        flags=re.IGNORECASE
    )

    if match:
        return int(
            match.group(1)
        )

    return None


def infer_fidelity(path):

    text = str(path).lower()

    if "high fidelity" in text:
        return "HIGH_FIDELITY"

    if "low fidelity" in text:
        return "LOW_FIDELITY"

    return None
```

## 9. Normalizar URLs para enlazar DOCX ↔ metadata

```python
URL_RE = re.compile(
    r"https?://[^\s<>\[\]\(\){}\"']+",
    flags=re.IGNORECASE
)


def clean_url_token(value):

    if value is None:
        return None

    value = (
        str(value)
        .strip()
        .replace("\u200b", "")
        .replace("\u202f", "")
        .replace("\xa0", "")
    )

    value = value.rstrip(
        ".,;:!?)]}>"
    )

    return value


def normalize_source_url(value):

    if pd.isna(value):
        return None

    value = clean_url_token(value)

    try:

        parsed = urlparse(value)

        host = (
            parsed.netloc
            .lower()
            .replace(
                "www.",
                ""
            )
        )

        path = unquote(
            parsed.path
        ).rstrip("/")

        return (
            host
            +
            path.lower()
        )

    except Exception:
        return None


metadata_url_registry = (
    metadata[
        [
            "Prototype",
            "Prototype Type",
            "Domain",
            "Scource URL"
        ]
    ]
    .copy()
)

metadata_url_registry[
    "normalized_url"
] = metadata_url_registry[
    "Scource URL"
].apply(
    normalize_source_url
)

display(
    metadata_url_registry
)
```

```python
metadata_url_map = defaultdict(list)

for _, row in metadata_url_registry.iterrows():

    key = row[
        "normalized_url"
    ]

    if key is None:
        continue

    metadata_url_map[
        key
    ].append(
        row[
            "Prototype"
        ]
    )
```

## 10. Construir registro de DOCX y enlazar por URL

```python
doc_registry_records = []

doc_text_cache = {}

for path in docx_files:

    text = extract_docx_text(
        path
    )

    doc_text_cache[
        str(path)
    ] = text

    urls = [
        clean_url_token(x)
        for x in URL_RE.findall(
            text
        )
    ]

    normalized_urls = []

    for url in urls:

        normalized = normalize_source_url(
            url
        )

        if normalized:
            normalized_urls.append(
                normalized
            )

    matched_prototypes = []

    for normalized in normalized_urls:

        matched_prototypes.extend(
            metadata_url_map.get(
                normalized,
                []
            )
        )

    matched_prototypes = sorted(
        set(
            matched_prototypes
        )
    )

    if len(matched_prototypes) == 1:

        match_status = "MATCHED_BY_URL"
        selected_prototype = matched_prototypes[0]

    elif len(matched_prototypes) > 1:

        match_status = "AMBIGUOUS_URL_MATCH"
        selected_prototype = None

    else:

        match_status = "NO_URL_MATCH"
        selected_prototype = None

    doc_registry_records.append({
        "file":
            str(
                path.relative_to(
                    VT_DIR
                )
            ),

        "document_name":
            path.stem,

        "evaluation_round":
            infer_evaluation_round(
                path
            ),

        "fidelity":
            infer_fidelity(
                path
            ),

        "urls_found":
            len(
                normalized_urls
            ),

        "normalized_urls":
            " | ".join(
                sorted(
                    set(
                        normalized_urls
                    )
                )
            ),

        "match_status":
            match_status,

        "prototype":
            selected_prototype
    })

doc_registry = pd.DataFrame(
    doc_registry_records
)

print(
    doc_registry[
        "match_status"
    ].value_counts(
        dropna=False
    )
)

display(
    doc_registry
)
```

## 11. Extraer secciones de H1, H4, H5 y H6

Se usan los nombres completos de Nielsen como delimitadores. La extracción es evidencia forense del documento; no es X del modelo.


```python
HEURISTICS = {
    "H1":
        "Visibility of System Status",

    "H4":
        "Consistency and Standards",

    "H5":
        "Error Prevention",

    "H6":
        "Recognition Rather Than Recall"
}

ALL_HEURISTIC_HEADINGS = [
    "Visibility of System Status",
    "Match Between the System and the Real World",
    "User Control and Freedom",
    "Consistency and Standards",
    "Error Prevention",
    "Recognition Rather Than Recall",
    "Flexibility and Efficiency of Use",
    "Aesthetic and Minimalist Design",
    "Help Users Recognize, Diagnose, and Recover from Errors",
    "Help and Documentation"
]


def normalize_spaces(text):

    return re.sub(
        r"\s+",
        " ",
        str(text)
    ).strip()


def extract_heuristic_section(
    text,
    heading
):

    normalized = normalize_spaces(
        text
    )

    start_match = re.search(
        re.escape(
            heading
        ),
        normalized,
        flags=re.IGNORECASE
    )

    if not start_match:
        return None

    start = start_match.start()

    candidate_ends = []

    for other in ALL_HEURISTIC_HEADINGS:

        if other.lower() == heading.lower():
            continue

        match = re.search(
            re.escape(
                other
            ),
            normalized[
                start_match.end():
            ],
            flags=re.IGNORECASE
        )

        if match:

            candidate_ends.append(
                start_match.end()
                +
                match.start()
            )

    end = (
        min(
            candidate_ends
        )
        if candidate_ends
        else len(
            normalized
        )
    )

    return normalized[
        start:end
    ]
```

## 12. Detectar evidencia de severidad y ausencia explícita de problemas

```python
SEVERITY_PATTERNS = [
    re.compile(
        r"Severity\s*(?:Rating)?\s*[:\-]?\s*([0-4])(?:\.0)?\b",
        flags=re.IGNORECASE
    ),
    re.compile(
        r"Severity\s*[:\-]?\s*([0-4])\s*\(",
        flags=re.IGNORECASE
    )
]

OVERALL_PATTERNS = [
    re.compile(
        r"Overall\s+(?:Severity\s+)?Rating\s*[:\-]\s*([0-4])(?:\.0)?\b",
        flags=re.IGNORECASE
    ),
    re.compile(
        r"Overall\s+Rating\s*(?:is)?\s*[:\-]?\s*([0-4])(?:\.0)?\b",
        flags=re.IGNORECASE
    )
]

NO_ISSUE_PATTERNS = [
    re.compile(
        r"no\s+(?:major\s+)?(?:usability\s+)?issues?\s+(?:were\s+)?found",
        flags=re.IGNORECASE
    ),
    re.compile(
        r"no\s+issues?\s+identified",
        flags=re.IGNORECASE
    ),
    re.compile(
        r"no\s+violations?\s+(?:were\s+)?found",
        flags=re.IGNORECASE
    ),
    re.compile(
        r"no\s+problem",
        flags=re.IGNORECASE
    )
]


def collect_numeric_matches(
    text,
    patterns
):

    values = []

    for pattern in patterns:

        for match in pattern.finditer(
            text
        ):

            try:
                values.append(
                    int(
                        match.group(1)
                    )
                )
            except Exception:
                pass

    return values


def explicit_no_issue(
    text
):

    return any(
        pattern.search(
            text
        )
        is not None
        for pattern
        in NO_ISSUE_PATTERNS
    )
```

## 13. Construir evidencia por documento y heurística

Regla conservadora del candidato binario:

- `1`: existe al menos una severidad explícita > 0;
- `0`: no hay severidad > 0 y el texto dice explícitamente que no se encontraron problemas;
- `NaN`: evidencia ambigua o insuficiente.

No se usa `H* Compare` para construir estas etiquetas.


```python
heuristic_evidence_records = []

for _, doc_row in doc_registry.iterrows():

    file_rel = doc_row[
        "file"
    ]

    path = VT_DIR / file_rel

    text = doc_text_cache.get(
        str(path),
        ""
    )

    for heuristic_code, heading in (
        HEURISTICS.items()
    ):

        section = extract_heuristic_section(
            text,
            heading
        )

        if not section:

            heuristic_evidence_records.append({
                "file":
                    file_rel,

                "evaluation_round":
                    doc_row[
                        "evaluation_round"
                    ],

                "prototype":
                    doc_row[
                        "prototype"
                    ],

                "match_status":
                    doc_row[
                        "match_status"
                    ],

                "heuristic":
                    heuristic_code,

                "section_found":
                    False,

                "severity_values":
                    None,

                "overall_rating_values":
                    None,

                "explicit_no_issue":
                    False,

                "label_candidate":
                    np.nan,

                "label_basis":
                    "SECTION_NOT_FOUND"
            })

            continue

        severity_values = collect_numeric_matches(
            section,
            SEVERITY_PATTERNS
        )

        overall_values = collect_numeric_matches(
            section,
            OVERALL_PATTERNS
        )

        no_issue = explicit_no_issue(
            section
        )

        positive_severity = [
            value
            for value
            in severity_values
            if value > 0
        ]

        if positive_severity:

            label_candidate = 1
            label_basis = "EXPLICIT_POSITIVE_SEVERITY"

        elif no_issue:

            label_candidate = 0
            label_basis = "EXPLICIT_NO_ISSUE"

        else:

            label_candidate = np.nan
            label_basis = "AMBIGUOUS_TEXT"

        heuristic_evidence_records.append({
            "file":
                file_rel,

            "evaluation_round":
                doc_row[
                    "evaluation_round"
                ],

            "prototype":
                doc_row[
                    "prototype"
                ],

            "match_status":
                doc_row[
                    "match_status"
                ],

            "heuristic":
                heuristic_code,

            "section_found":
                True,

            "severity_values":
                str(
                    severity_values
                ),

            "overall_rating_values":
                str(
                    overall_values
                ),

            "explicit_no_issue":
                no_issue,

            "label_candidate":
                label_candidate,

            "label_basis":
                label_basis
        })

heuristic_evidence = pd.DataFrame(
    heuristic_evidence_records
)

print(
    "Evidence rows:",
    heuristic_evidence.shape
)

print(
    "\nSection found:"
)

print(
    heuristic_evidence[
        "section_found"
    ].value_counts(
        dropna=False
    )
)

print(
    "\nLabel basis:"
)

print(
    heuristic_evidence[
        "label_basis"
    ].value_counts(
        dropna=False
    )
)

display(
    heuristic_evidence.head(
        100
    )
)
```

## 15. Consolidación provisional por prototipo y heurística

Solo se consolida cuando existen etiquetas candidatas no nulas. El resultado conserva número de evaluaciones y acuerdo; todavía se denomina `candidate`, no Y final.


```python
def consolidate_candidate_labels(
    group
):

    valid = (
        group[
            "label_candidate"
        ]
        .dropna()
        .astype(int)
        .tolist()
    )

    if not valid:

        return pd.Series({
            "n_valid_evaluations":
                0,

            "n_positive":
                0,

            "n_negative":
                0,

            "agreement_ratio":
                np.nan,

            "consensus_candidate":
                np.nan,

            "consensus_status":
                "NO_VALID_LABEL"
        })

    n_positive = sum(
        value == 1
        for value
        in valid
    )

    n_negative = sum(
        value == 0
        for value
        in valid
    )

    majority = max(
        n_positive,
        n_negative
    )

    agreement_ratio = (
        majority
        /
        len(valid)
    )

    if n_positive > n_negative:

        consensus = 1
        status = "MAJORITY_POSITIVE"

    elif n_negative > n_positive:

        consensus = 0
        status = "MAJORITY_NEGATIVE"

    else:

        consensus = np.nan
        status = "TIE"

    return pd.Series({
        "n_valid_evaluations":
            len(valid),

        "n_positive":
            n_positive,

        "n_negative":
            n_negative,

        "agreement_ratio":
            agreement_ratio,

        "consensus_candidate":
            consensus,

        "consensus_status":
            status
    })


candidate_consensus_long = (
    matched_label_evidence
    .groupby(
        [
            "prototype",
            "heuristic"
        ],
        group_keys=False
    )
    .apply(
        consolidate_candidate_labels,
        include_groups=False
    )
    .reset_index()
)

display(
    candidate_consensus_long
)
```

```python
candidate_wide = (
    candidate_consensus_long
    .pivot(
        index="prototype",
        columns="heuristic",
        values="consensus_candidate"
    )
    .reset_index()
)

candidate_wide.columns.name = None

candidate_wide = candidate_wide.rename(
    columns={
        "H1":
            "H1_candidate",

        "H4":
            "H4_candidate",

        "H5":
            "H5_candidate",

        "H6":
            "H6_candidate"
    }
)

print(
    "Candidate wide:",
    candidate_wide.shape
)

display(
    candidate_wide
)
```
