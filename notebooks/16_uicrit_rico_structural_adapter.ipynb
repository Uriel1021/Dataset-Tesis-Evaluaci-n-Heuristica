```python
from pathlib import Path

import pandas as pd
import numpy as np
import json
import tarfile

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
pd.set_option("display.max_colwidth", 300)

PROJECT_DIR = Path("..").resolve()

PUBLIC_RAW = (
    PROJECT_DIR
    / "data"
    / "public_v2"
    / "raw"
)

INTERIM_DIR = (
    PROJECT_DIR
    / "data"
    / "public_v2"
    / "interim"
)

REPORTS_DIR = (
    PROJECT_DIR
    / "reports"
    / "tables"
)

RICO_DIR = (
    PUBLIC_RAW
    / "10_uicrit_rico"
)

INTERIM_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
uicrit["rico_id_num"] = pd.to_numeric(
    uicrit["rico_id"],
    errors="coerce"
).astype("Int64")

ui_details["rico_id_num"] = pd.to_numeric(
    ui_details["UI Number"],
    errors="coerce"
).astype("Int64")
```

```python
target_rico_ids = set(
    uicrit[
        "rico_id_num"
    ]
    .dropna()
    .astype(int)
)

rico_available_ids = set(
    ui_details[
        "rico_id_num"
    ]
    .dropna()
    .astype(int)
)

overlap = (
    target_rico_ids
    &
    rico_available_ids
)

print(
    "UICrit IDs:",
    len(target_rico_ids)
)

print(
    "Encontrados en RICO:",
    len(overlap)
)

print(
    "Cobertura:",
    len(overlap)
    / len(target_rico_ids)
)
```

```python
archive_path = (
    RICO_DIR
    / "unique_uis.tar.gz"
)

print(
    "Archivo:",
    archive_path.name
)

print(
    "GB:",
    archive_path.stat().st_size
    / 1024**3
)
```

```python
selected_hierarchies = {}
selected_member_names = []

with tarfile.open(
    archive_path,
    mode="r:gz"
) as tar:

    for member in tar:

        if not member.isfile():
            continue

        path = Path(
            member.name
        )

        if (
            path.suffix.lower()
            != ".json"
        ):
            continue

        try:
            rico_id = int(
                path.stem
            )
        except ValueError:
            continue

        if rico_id not in target_rico_ids:
            continue

        extracted = tar.extractfile(
            member
        )

        if extracted is None:
            continue

        try:
            obj = json.load(
                extracted
            )
        except Exception as e:

            print(
                "Error JSON:",
                rico_id,
                e
            )

            continue

        selected_hierarchies[
            rico_id
        ] = obj

        selected_member_names.append(
            member.name
        )

        if (
            len(selected_hierarchies)
            ==
            len(target_rico_ids)
        ):
            break
```

```python
from collections import Counter
```

```python
def get_root(obj):

    if not isinstance(obj, dict):
        return None

    activity = obj.get(
        "activity"
    )

    if not isinstance(
        activity,
        dict
    ):
        return None

    return activity.get(
        "root"
    )
```

```python
def walk_nodes(
    node,
    depth=0
):

    if not isinstance(
        node,
        dict
    ):
        return

    yield node, depth

    children = node.get(
        "children",
        []
    )

    if not isinstance(
        children,
        list
    ):
        return

    for child in children:

        yield from walk_nodes(
            child,
            depth + 1
        )
```

```python
node_key_counter = Counter()

node_count_total = 0

for rico_id, obj in (
    selected_hierarchies.items()
):

    root = get_root(
        obj
    )

    if root is None:
        continue

    for node, depth in walk_nodes(
        root
    ):

        node_count_total += 1

        node_key_counter.update(
            node.keys()
        )
```

```python
node_schema = pd.DataFrame(
    node_key_counter.most_common(),
    columns=[
        "node_property",
        "count"
    ]
)

node_schema[
    "coverage_ratio"
] = (
    node_schema["count"]
    /
    node_count_total
)

print(
    "Nodos totales:",
    node_count_total
)

display(
    node_schema.head(50)
)
```

```python
text_candidate_fields = [
    "text",
    "text-content",
    "textContent",
    "content-desc",
    "resource-id"
]

text_field_audit = []

for field in text_candidate_fields:

    present = 0
    nonempty = 0

    for rico_id, obj in (
        selected_hierarchies.items()
    ):

        root = get_root(
            obj
        )

        if root is None:
            continue

        for node, _ in walk_nodes(
            root
        ):

            if field not in node:
                continue

            present += 1

            value = node.get(
                field
            )

            if isinstance(
                value,
                list
            ):

                valid = any(
                    x is not None
                    and
                    str(x).strip()
                    not in {
                        "",
                        "null",
                        "None"
                    }
                    for x in value
                )

            else:

                valid = (
                    value is not None
                    and
                    str(value).strip()
                    not in {
                        "",
                        "null",
                        "None"
                    }
                )

            if valid:
                nonempty += 1

    text_field_audit.append({
        "field":
            field,

        "present":
            present,

        "nonempty":
            nonempty
    })

text_field_audit = pd.DataFrame(
    text_field_audit
)

text_field_audit
```

```python
def has_nonempty_value(
    value
):

    if value is None:
        return False

    if isinstance(
        value,
        list
    ):

        return any(
            has_nonempty_value(x)
            for x in value
        )

    text = str(
        value
    ).strip()

    return (
        text != ""
        and
        text.lower()
        not in {
            "none",
            "null",
            "nan"
        }
    )
```

```python
def is_visible_node(
    node
):

    visible_to_user = node.get(
        "visible-to-user"
    )

    visibility = node.get(
        "visibility"
    )

    if (
        visible_to_user
        is False
    ):
        return False

    if (
        isinstance(
            visibility,
            str
        )
        and
        visibility.lower()
        in {
            "gone",
            "invisible"
        }
    ):
        return False

    return True
```

```python
def is_interactive_node(
    node
):

    return any([
        node.get(
            "clickable"
        ) is True,

        node.get(
            "focusable"
        ) is True,

        node.get(
            "long-clickable"
        ) is True,

        node.get(
            "scrollable-vertical"
        ) is True,

        node.get(
            "scrollable-horizontal"
        ) is True
    ])
```

```python
def has_label_proxy(
    node
):

    text_fields = [
        "text",
        "text-content",
        "textContent",
        "content-desc"
    ]

    return any(
        has_nonempty_value(
            node.get(field)
        )
        for field in text_fields
    )
```

```python
def extract_rico_features(
    rico_id,
    obj
):

    root = get_root(
        obj
    )

    if root is None:

        return {
            "rico_id":
                rico_id,

            "rico_parse_ok":
                False
        }

    nodes = list(
        walk_nodes(
            root
        )
    )

    node_objects = [
        node
        for node, depth
        in nodes
    ]

    depths = [
        depth
        for node, depth
        in nodes
    ]

    visible_nodes = [
        node
        for node
        in node_objects
        if is_visible_node(
            node
        )
    ]

    interactive_nodes = [
        node
        for node
        in visible_nodes
        if is_interactive_node(
            node
        )
    ]

    labeled_interactive = [
        node
        for node
        in interactive_nodes
        if has_label_proxy(
            node
        )
    ]

    leaves = [
        node
        for node
        in node_objects
        if not isinstance(
            node.get(
                "children"
            ),
            list
        )
        or
        len(
            node.get(
                "children",
                []
            )
        )
        == 0
    ]

    internal_nodes = [
        node
        for node
        in node_objects
        if isinstance(
            node.get(
                "children"
            ),
            list
        )
        and
        len(
            node.get(
                "children"
            )
        )
        > 0
    ]

    branching_values = [
        len(
            node.get(
                "children",
                []
            )
        )
        for node
        in internal_nodes
    ]

    resource_id_count = sum(
        has_nonempty_value(
            node.get(
                "resource-id"
            )
        )
        for node
        in visible_nodes
    )

    class_values = [
        str(
            node.get(
                "class"
            )
        )
        for node
        in visible_nodes
        if has_nonempty_value(
            node.get(
                "class"
            )
        )
    ]

    scrollable_count = sum(
        (
            node.get(
                "scrollable-vertical"
            )
            is True
        )
        or
        (
            node.get(
                "scrollable-horizontal"
            )
            is True
        )
        for node
        in visible_nodes
    )

    return {
        "rico_id":
            rico_id,

        "rico_parse_ok":
            True,

        "rico_node_count":
            len(
                node_objects
            ),

        "rico_visible_node_count":
            len(
                visible_nodes
            ),

        "rico_leaf_count":
            len(
                leaves
            ),

        "rico_max_tree_depth":
            (
                max(
                    depths
                )
                if depths
                else 0
            ),

        "rico_mean_branching_factor":
            (
                np.mean(
                    branching_values
                )
                if branching_values
                else 0
            ),

        "rico_interactive_node_count":
            len(
                interactive_nodes
            ),

        "rico_scrollable_node_count":
            scrollable_count,

        "rico_resource_id_ratio":
            (
                resource_id_count
                /
                len(
                    visible_nodes
                )
                if visible_nodes
                else np.nan
            ),

        "rico_interactive_label_proxy_ratio":
            (
                len(
                    labeled_interactive
                )
                /
                len(
                    interactive_nodes
                )
                if interactive_nodes
                else np.nan
            ),

        "rico_unlabeled_interactive_count":
            (
                len(
                    interactive_nodes
                )
                -
                len(
                    labeled_interactive
                )
            ),

        "rico_unique_class_count":
            len(
                set(
                    class_values
                )
            ),

        "rico_class_diversity_ratio":
            (
                len(
                    set(
                        class_values
                    )
                )
                /
                len(
                    class_values
                )
                if class_values
                else np.nan
            )
    }
```

```python
rico_feature_records = []

for rico_id, obj in (
    selected_hierarchies.items()
):

    rico_feature_records.append(
        extract_rico_features(
            rico_id,
            obj
        )
    )

rico_features = pd.DataFrame(
    rico_feature_records
)
```

```python
uicrit_proxy_targets = (
    uicrit
    .groupby(
        "rico_id_num"
    )
    .agg(
        proxy_record_count=(
            "rico_id",
            "size"
        ),

        learnability_proxy=(
            "learnability",
            "median"
        ),

        design_quality_proxy=(
            "design_quality_rating",
            "median"
        ),

        learnability_proxy_mean=(
            "learnability",
            "mean"
        ),

        design_quality_proxy_mean=(
            "design_quality_rating",
            "mean"
        )
    )
    .reset_index()
    .rename(
        columns={
            "rico_id_num":
                "rico_id"
        }
    )
)
```
