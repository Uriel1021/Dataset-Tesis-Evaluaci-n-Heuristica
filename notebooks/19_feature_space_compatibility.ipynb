```python
declared_mappings = pd.DataFrame(
    mapping_records
)

display(
    declared_mappings
)
```

```python
canonical_names = set(
    canonical_x[
        "variable"
    ]
)

rico_names = set(
    rico_features.columns
)

catching_names = set(
    catching_features.columns
)
```

```python
for _, row in declared_mappings.iterrows():

    assert (
        row[
            "canonical_variable"
        ]
        in canonical_names
    ), row[
        "canonical_variable"
    ]

    if (
        row[
            "source_dataset"
        ]
        ==
        "UICrit_RICO"
    ):

        assert (
            row[
                "source_variable"
            ]
            in rico_names
        )

    if (
        row[
            "source_dataset"
        ]
        ==
        "Catching_UX_Flaws"
    ):

        assert (
            row[
                "source_variable"
            ]
            in catching_names
        )

print(
    "Mappings validados ✅"
)
```

```python
full_records = []

for source in [
    "UICrit_RICO",
    "Catching_UX_Flaws"
]:

    source_mapping = (
        declared_mappings[
            declared_mappings[
                "source_dataset"
            ]
            ==
            source
        ]
    )

    for canonical in (
        canonical_x[
            "variable"
        ]
    ):

        matches = (
            source_mapping[
                source_mapping[
                    "canonical_variable"
                ]
                ==
                canonical
            ]
        )

        if matches.empty:

            full_records.append({
                "source_dataset":
                    source,

                "canonical_variable":
                    canonical,

                "source_variable":
                    None,

                "compatibility_level":
                    "NO_MAPPING",

                "possible_heuristic":
                    None,

                "reason":
                    "No se encontró una operacionalización compatible en esta fuente."
            })

        else:

            for _, mapping in (
                matches.iterrows()
            ):

                full_records.append(
                    mapping.to_dict()
                )

feature_compatibility_matrix = pd.DataFrame(
    full_records
)
```

```python
feature_compatibility_matrix = (
    feature_compatibility_matrix
    .merge(
        compatibility_levels[
            [
                "compatibility_level",
                "can_be_renamed_canonical",
                "can_share_feature_space"
            ]
        ],
        on="compatibility_level",
        how="left",
        validate="many_to_one"
    )
)
```

```python
display(
    feature_compatibility_matrix
)
```

```python
compatibility_summary = (
    feature_compatibility_matrix
    .groupby(
        [
            "source_dataset",
            "compatibility_level"
        ]
    )
    .size()
    .reset_index(
        name="count"
    )
)

display(
    compatibility_summary
)
```

```python
exact_features = (
    feature_compatibility_matrix[
        feature_compatibility_matrix[
            "compatibility_level"
        ]
        ==
        "EXACT"
    ]
)
```

```python
exact_per_source = (
    exact_features
    .groupby(
        "source_dataset"
    )[
        "canonical_variable"
    ]
    .apply(set)
    .to_dict()
)

rico_exact = exact_per_source.get(
    "UICrit_RICO",
    set()
)

catching_exact = exact_per_source.get(
    "Catching_UX_Flaws",
    set()
)

strict_common_space = (
    rico_exact
    &
    catching_exact
)

print(
    "EXACT RICO:",
    len(
        rico_exact
    )
)

print(
    "EXACT Catching:",
    len(
        catching_exact
    )
)

print(
    "Espacio común EXACT:",
    len(
        strict_common_space
    )
)

print(
    sorted(
        strict_common_space
    )
)
```
