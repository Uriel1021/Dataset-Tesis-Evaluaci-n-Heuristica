```python
workbook_path = (
    DASH_DIR
    / "Interaction Patterns Dataset.xlsx"
)

xls = pd.ExcelFile(
    workbook_path
)

print(
    xls.sheet_names
)
```

```python
for sheet in xls.sheet_names:

    print(
        "\n",
        "=" * 100
    )

    print(
        "SHEET:",
        sheet
    )

    raw_sheet = pd.read_excel(
        workbook_path,
        sheet_name=sheet,
        header=None,
        dtype=str,
        nrows=15
    )

    print(
        "Shape preview:",
        raw_sheet.shape
    )

    display(
        raw_sheet
    )
```

```python
participant_ids = set(
    dash_main[
        "P"
    ]
    .dropna()
    .astype(str)
)

participant_matches = []

for sheet in xls.sheet_names:

    temp = pd.read_excel(
        workbook_path,
        sheet_name=sheet,
        header=None,
        dtype=str
    )

    for row_idx in range(
        len(temp)
    ):

        for col_idx in range(
            temp.shape[1]
        ):

            value = temp.iat[
                row_idx,
                col_idx
            ]

            if (
                pd.notna(value)
                and
                str(value).strip()
                in participant_ids
            ):

                participant_matches.append({
                    "sheet":
                        sheet,

                    "row":
                        row_idx,

                    "column":
                        col_idx,

                    "participant":
                        str(value).strip()
                })
```

```python
participant_match_df = pd.DataFrame(
    participant_matches
)

print(
    "Coincidencias de participantes:",
    len(
        participant_match_df
    )
)

if not participant_match_df.empty:

    print(
        "Participantes distintos encontrados:",
        participant_match_df[
            "participant"
        ].nunique()
    )

    display(
        participant_match_df.head(100)
    )
```

```python
dash_zip = (
    DASH_DIR
    / "Interaction Data.zip"
)

with zipfile.ZipFile(
    dash_zip,
    "r"
) as z:

    real_jsons = [
        name
        for name in z.namelist()
        if (
            name.endswith(".json")
            and
            "__MACOSX"
            not in name
            and
            not Path(name).name.startswith(
                "._"
            )
        )
    ]

print(
    "Logs reales:",
    len(real_jsons)
)
```

```python
log_registry = pd.DataFrame({
    "source_file":
        [
            Path(name).name
            for name in real_jsons
        ]
})

log_registry[
    "sessionstart_from_filename"
] = pd.to_numeric(
    log_registry[
        "source_file"
    ]
    .str.replace(
        ".json",
        "",
        regex=False
    ),
    errors="coerce"
)

display(
    log_registry.head(20)
)

print(
    log_registry.shape
)
```

```python
session_ids = set(
    log_registry[
        "sessionstart_from_filename"
    ]
    .dropna()
    .astype("int64")
    .astype(str)
)

session_matches = []

for sheet in xls.sheet_names:

    temp = pd.read_excel(
        workbook_path,
        sheet_name=sheet,
        header=None,
        dtype=str
    )

    for row_idx in range(
        len(temp)
    ):

        for col_idx in range(
            temp.shape[1]
        ):

            value = temp.iat[
                row_idx,
                col_idx
            ]

            if pd.isna(value):
                continue

            clean_value = (
                str(value)
                .strip()
                .replace(
                    ".0",
                    ""
                )
            )

            if clean_value in session_ids:

                session_matches.append({
                    "sheet":
                        sheet,

                    "row":
                        row_idx,

                    "column":
                        col_idx,

                    "sessionstartms":
                        clean_value
                })
```

```python
session_match_df = pd.DataFrame(
    session_matches
)

print(
    "Session IDs encontrados en workbook:",
    len(
        session_match_df
    )
)

if not session_match_df.empty:

    print(
        "Sesiones distintas:",
        session_match_df[
            "sessionstartms"
        ].nunique()
    )

    display(
        session_match_df.head(100)
    )
```

```python
participant_log_matches = []

with zipfile.ZipFile(
    dash_zip,
    "r"
) as z:

    for name in real_jsons:

        raw = z.read(
            name
        )

        text = raw.decode(
            "utf-8",
            errors="ignore"
        )

        found = [
            participant
            for participant
            in participant_ids
            if participant in text
        ]

        for participant in found:

            participant_log_matches.append({
                "source_file":
                    Path(name).name,

                "participant":
                    participant
            })
```

```python
participant_log_match_df = pd.DataFrame(
    participant_log_matches
)

print(
    "Matches P dentro de logs:",
    len(
        participant_log_match_df
    )
)

if not participant_log_match_df.empty:

    display(
        participant_log_match_df
    )
```

```python
join_audit = pd.DataFrame([
    {
        "test":
            "P_IN_WORKBOOK",

        "matches":
            len(
                participant_match_df
            ),

        "unique_matches":
            (
                participant_match_df[
                    "participant"
                ].nunique()
                if not participant_match_df.empty
                else 0
            )
    },

    {
        "test":
            "SESSIONSTART_IN_WORKBOOK",

        "matches":
            len(
                session_match_df
            ),

        "unique_matches":
            (
                session_match_df[
                    "sessionstartms"
                ].nunique()
                if not session_match_df.empty
                else 0
            )
    },

    {
        "test":
            "P_IN_RAW_LOGS",

        "matches":
            len(
                participant_log_match_df
            ),

        "unique_matches":
            (
                participant_log_match_df[
                    "participant"
                ].nunique()
                if not participant_log_match_df.empty
                else 0
            )
    }
])

join_audit
```

```python
OBJECT_ID_RE = re.compile(
    r'ObjectId\("([^"]+)"\)'
)

ISO_DATE_RE = re.compile(
    r'ISODate\("([^"]+)"\)'
)

NUMBER_LONG_RE = re.compile(
    r'NumberLong\("?(-?\d+)"?\)'
)


def parse_mongo_line(line):

    line = line.strip()

    if not line:
        return None

    line = OBJECT_ID_RE.sub(
        r'"\1"',
        line
    )

    line = ISO_DATE_RE.sub(
        r'"\1"',
        line
    )

    line = NUMBER_LONG_RE.sub(
        r'\1',
        line
    )

    return json.loads(
        line
    )
```

```python
event_samples = []

wanted_events = {
    "mouseup",
    "mousedown",
    "keypress",
    "keydown",
    "scroll",
    "change",
    "dblclick"
}

with zipfile.ZipFile(
    dash_zip,
    "r"
) as z:

    for name in real_jsons[:5]:

        with z.open(name) as f:

            for raw_line in f:

                text = raw_line.decode(
                    "utf-8-sig",
                    errors="replace"
                )

                obj = parse_mongo_line(
                    text
                )

                if obj is None:
                    continue

                event = obj.get(
                    "event"
                )

                if event not in wanted_events:
                    continue

                event_samples.append({
                    "source_file":
                        Path(name).name,

                    "event":
                        event,

                    "timestampms":
                        obj.get(
                            "timestampms"
                        ),

                    "episodeCount":
                        obj.get(
                            "episodeCount"
                        ),

                    "urlFull":
                        obj.get(
                            "urlFull"
                        ),

                    "mouse":
                        obj.get(
                            "mouse"
                        ),

                    "key":
                        obj.get(
                            "key"
                        ),

                    "scroll":
                        obj.get(
                            "scroll"
                        ),

                    "node":
                        obj.get(
                            "node"
                        ),

                    "change":
                        obj.get(
                            "change"
                        )
                })

                if len(
                    event_samples
                ) >= 100:
                    break

        if len(
            event_samples
        ) >= 100:
            break
```

```python
event_sample_df = pd.DataFrame(
    event_samples
)

display(
    event_sample_df.head(100)
)
```
