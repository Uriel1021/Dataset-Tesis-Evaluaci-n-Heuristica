```python
def inspeccionar_csv(path, nrows=5):

    try:
        df = pd.read_csv(
            path,
            nrows=nrows,
            low_memory=False
        )

        return {
            "file": str(
                path.relative_to(PUBLIC_RAW)
            ),
            "n_columns": len(df.columns),
            "columns": list(df.columns),
            "status": "OK"
        }

    except Exception as e:

        return {
            "file": str(
                path.relative_to(PUBLIC_RAW)
            ),
            "n_columns": np.nan,
            "columns": None,
            "status": f"ERROR: {e}"
        }
```

```python
def inspeccionar_json(path):

    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:
            data = json.load(f)

        if isinstance(data, dict):

            structure = "dict"
            keys = list(data.keys())

        elif isinstance(data, list):

            structure = "list"

            if (
                len(data) > 0
                and
                isinstance(data[0], dict)
            ):
                keys = list(data[0].keys())
            else:
                keys = []

        else:
            structure = type(data).__name__
            keys = []

        return {
            "file": str(
                path.relative_to(PUBLIC_RAW)
            ),
            "structure": structure,
            "top_level_keys": keys,
            "status": "OK"
        }

    except Exception as e:

        return {
            "file": str(
                path.relative_to(PUBLIC_RAW)
            ),
            "structure": None,
            "top_level_keys": None,
            "status": f"ERROR: {e}"
        }
```

```python
import re

OBJECT_ID_RE = re.compile(
    r'ObjectId\("([^"]+)"\)'
)

ISO_DATE_RE = re.compile(
    r'ISODate\("([^"]+)"\)'
)

NUMBER_LONG_RE = re.compile(
    r'NumberLong\("?(-?\d+)"?\)'
)


def parse_mongo_export_line(line):

    line = line.strip()

    if not line:
        return None

    # MongoDB ObjectId -> string JSON
    line = OBJECT_ID_RE.sub(
        r'"\1"',
        line
    )

    # Por si aparecen fechas MongoDB
    line = ISO_DATE_RE.sub(
        r'"\1"',
        line
    )

    # Por si aparecen NumberLong
    line = NUMBER_LONG_RE.sub(
        r'\1',
        line
    )

    return json.loads(line)
```

```python
dash_event_records = []
dash_parse_errors = []
dash_keys = set()

with zipfile.ZipFile(
    dash_zip,
    "r"
) as z:

    for name in real_jsons:

        with z.open(name) as f:

            for line_number, raw_line in enumerate(
                f,
                start=1
            ):

                text = raw_line.decode(
                    "utf-8-sig",
                    errors="replace"
                )

                try:

                    obj = parse_mongo_export_line(
                        text
                    )

                    if obj is None:
                        continue

                    dash_keys.update(
                        obj.keys()
                    )

                    dash_event_records.append({
                        "source_file": name,
                        "line_number": line_number,
                        "event": obj.get("event"),
                        "timestampms": obj.get("timestampms"),
                        "sessionstartms": obj.get("sessionstartms"),
                        "episodeCount": obj.get("episodeCount"),
                        "sd": obj.get("sd"),
                        "sid": obj.get("sid"),
                        "url": obj.get("url"),
                        "urlFull": obj.get("urlFull")
                    })

                except Exception as e:

                    dash_parse_errors.append({
                        "source_file": name,
                        "line_number": line_number,
                        "error": str(e),
                        "sample": text[:200]
                    })
```

```python
dash_events = pd.DataFrame(
    dash_event_records
)

dash_errors = pd.DataFrame(
    dash_parse_errors
)

print(
    "Eventos parseados:",
    len(dash_events)
)

print(
    "Errores:",
    len(dash_errors)
)

print(
    "Archivos:",
    dash_events[
        "source_file"
    ].nunique()
)

print(
    "\nKeys top-level encontradas:"
)

print(
    sorted(dash_keys)
)
```

```python
def normalize_heuristic(value):

    if pd.isna(value):
        return None

    text = str(value).strip().lower()

    # Eliminar prefijos como:
    # "1. Visibility..."
    text = re.sub(
        r"^\d+\.\s*",
        "",
        text
    )

    direct_map = {
        "visibility of system status": "H1",
        "consistency and standards": "H4",
        "error prevention": "H5",
        "recognition rather than recall": "H6"
    }

    if text in direct_map:
        return direct_map[text]

    # Variantes "Heuristic 1", etc.
    match = re.fullmatch(
        r"heuristic\s*(\d+)",
        text
    )

    if match:

        number = int(
            match.group(1)
        )

        if number in {
            1,
            4,
            5,
            6
        }:
            return f"H{number}"

    return None
```
